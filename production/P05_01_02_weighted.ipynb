```python
import pandas as pd
# S01, S02, S03 세 협력사의 불량률과 지연율
df = pd.DataFrame({'협력사': ['S01', 'S02', 'S03'],
                   '불량률': [0.5, 2.0, 1.0],      # 높을수록 위험
                   '지연율': [0.20, 0.05, 0.10]})  # 높을수록 위험

# 부여할 가중치 - 품질 리스크를 납기 지연 리스크보다 더 무겁게 반영
w = {'불량률': 0.6, '지연율': 0.4}

for k in w:
    # minMax 정규화로 0~1 사이 점수로 정규화
    df[k + '_n'] = (df[k] - df[k].min()) / (df[k].max() - df[k].min())

df # 정규화 한 내용
```

```python
# 정규화 점수를 합산한다.
# df['risk'] = sum(df[k + '_n'] * v for k,v in w.items()) * 100
df['risk'] = 0.0
for k, v in w.items():
    # 각 컬럼에 가중치를 곱한다음 합하여 risk 에 담는다.
    df['risk'] += df[k+'_n'] * v
df['risk'] = round(df['risk'] * 100,1) # 100점 만점 표현을 위해 100을 곱한다.
df.sort_values('risk')# 리스크다 낮은 순으로 보여준다.
```
