```python
from collections import Counter
from sklearn.datasets import make_classification
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score, recall_score

# 2000 개 중 95% 와 5% 비율로 데이터극 분리했다. 
x, y = make_classification(n_samples=2000, n_features=5, n_informative=3,
                           weights=[0.95, 0.05], random_state=0)
# 갯수 확인
print(Counter(y))

# 더미학습 - 무조건 빈도가 높은걸 정답이라고 말하도록 학습전략을 짠다.
dummy = DummyClassifier(strategy='most_frequent').fit(x, y)   
pred = dummy.predict(x)
print('정확도 :', accuracy_score(y, pred))     # 약 0.95
print('불량 재현율 :', recall_score(y, pred))  # 0 → 불량을 하나도 못 찾음
```
