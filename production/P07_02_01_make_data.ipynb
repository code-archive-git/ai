```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(7)
p, k = 20, 4
W = rng.normal(0, 1, (k, p))

def make(n):
    return rng.normal(0, 1, (n, k)) @ W + rng.normal(0, 0.3, (n, p))

X_normal = make(3000)
X_anom = make(40)
X_anom[:, :6] = X_anom[rng.permutation(40), :6]   # 값은 그대로, 관계만 뒤섞음

cols = [f'item_{i:02d}' for i in range(p)]
df = pd.DataFrame(np.vstack([X_normal, X_anom]), columns=cols)
df['lot_id'] = [f'L{i:05d}' for i in range(len(df))]
df['is_anomaly'] = [0] * 3000 + [1] * 40          # 평가용 (학습에는 사용 안 함)
df.sample(frac=1, random_state=0).to_csv('/content/lots.csv', index=False)
```
