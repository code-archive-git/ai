```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(1)
suppliers = [f'S{i:02d}' for i in range(1, 31)]
q_true = rng.beta(2, 40, 30)      # 숨은 불량 성향
d_true = rng.beta(2, 12, 30)      # 숨은 지연 성향
price  = rng.normal(100, 8, 30)   # 단가 지수 (싸고 빠른 업체 함정용)

rows = []
for s, q, d, p in zip(suppliers, q_true, d_true, price):
    for _ in range(rng.integers(15, 120)):               # 협력사별 로트 수가 다름
        late = rng.random() < d
        rows.append({
            'supplier'   : s,
            'price_idx'  : p,
            'n_insp'     : 50,                            # IQC 샘플 수
            'n_ng'       : rng.binomial(50, q),           # IQC 불량 수
            'delay_days' : int(rng.integers(1, 6)) if late else 0,
            'line_defect': int(rng.random() < q * 1.5),   # 투입 후 불량로트 여부
        })
lots = pd.DataFrame(rows)
lots.head()
```

```python
# Colab > Colab CPU > terminal
lots.to_csv('/content/data/supplier_lots.csv', index=False)
```
