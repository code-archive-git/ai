```python
import numpy as np
from scipy.stats import kurtosis
import matplotlib.pyplot as plt

rng = np.random.default_rng(0)
normal = rng.normal(0, 1, 2000) # 임의의 값 2000 개를 평균 0, 표준편차1 로 뽑아낸다.

# 이 데이터픞 복가사해서
impulse = normal.copy()
# impulse[::200] += 8 
impulse[0:len(normal):200] += 8 # 0~1999 까지 200단위 방번호 값에 8 더함

plt.figure(figsize=(14, 4))
plt.subplot(1,2,1)
plt.plot(normal, label="normal", c="green")
plt.legend()
plt.subplot(1,2,2)
plt.plot(impulse, label="impluse", c="green")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
for name, s in [('normal', normal), ('impulse', impulse)]:
    rms = np.sqrt(np.mean(s ** 2))
    peak = np.max(np.abs(s))
    print(f'{name:8s} RMS={rms:.2f} Peak={peak:.2f} Crest={peak/rms:.2f} Kurt={kurtosis(s):.2f}')
```
