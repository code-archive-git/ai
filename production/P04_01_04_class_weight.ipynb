```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from collections import Counter
```

```python
# 1. 불균형한 가상 데이터 생성 (정상 95%, 소수 5%)
x, y = make_classification(n_samples=2000, n_features=5, n_informative=3, weights=[0.95, 0.05], random_state=0)
# 2. 학습용과 테스트용 데이터 분리 (stratify=y로 클래스 비율 유지)
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, stratify=y, random_state=0)
```

```python
# 3. class_weight='balanced'가 설정된 RandomForest 모델 생성
model = RandomForestClassifier(
    class_weight='balanced', # class_weight 
    random_state=0
)

# 4. 교차 검증(Cross Validation) 설정 및 F1-Score 평가 수행
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=0)
scores = cross_val_score(model, x_train, y_train, cv=cv, scoring='f1')

print("\n--- 5-Fold 교차 검증 F1 점수들 ---")
print(scores)
print(f"교차 검증 평균 F1-Score: {scores.mean():.3f}")
```
