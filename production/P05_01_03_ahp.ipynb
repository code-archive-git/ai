```python
#- 지표를 2개씩 비교해 "A 가 B 보다 몇 배 중요한가"를 1~9점으로 답한다.
#- 비교표의 고유벡터로 가중치를 계산하고, CR(일관성 비율)이 0.1 미만이면 판단이 일관적이다.
```

```python
import numpy as np
# 이 부분을 현장인력과 기준을 정해야 한다.
#            품질  납기  단가
A = np.array([[1,   3,   5],    # 품질은 납기보다 3배, 단가보다 5배 중요
              [1/3, 1,   2],    # 납기는 품질의 1/3, 단가의 5/3=2배 중요(2가 아닌 5/3 으로 하면?)
              [1/5, 1/2, 1]])   # 단가는 품질의 1/5, 납기의 1/2 중요(1/(5/3) 으로 더 복잡해짐)
A
```

```python
# 행렬 A의 고윳값(val)과 고유벡터(vec)를 구합니다.
val, vec = np.linalg.eig(A) 
print(f'고유벡터:{vec}') # 여러 항목의 상대적 비중(품질,납기,단가)
""" 첫번째 컬럼이 실제적인 가중치 벡터
[0.92811934, -0.92811934,-0.92811934]
[0.32875799, 0.164379,   0.164379]
[0.17467875, 0.08733937, 0.08733937]
"""
# 행렬 A 안에서 얼마나 일관되게 증폭·반영되는지?
# 고유값 3개 [ 3.00369460,-1.84729903, -1.84729903]
# 0 : 최대 고윳값, 고윳값이 3개이니 3과 가까워야 한다. 일관성 비율에 사용됨
# 1, 2 : 사용하지 않음(앞의 3을 맞추기 위한 숫자들이라고 이해하지)
print(f'고유값:{val}') 
```

```python
# 어쨌든 val[0] 을 통해 CR 을 계산 하고
#  vec[i][0] 은 가중치 벡터라는 걸 알았다. 
w = vec[:,0] # vec 의 모든 행의 0번 인덱스 값을 뽑는다.
# print(w) # 그런데 불필요한 허수도 나오니...
w = np.real(w)
print(w) # real 을 통해 실수만 뽑아 낸다.
# 그리고 가중치 정규화
w /= w.sum() # 개별 값/w의 총합
print(w)
```

```python
# 고유값 에서 갯수(3)를 뺀다.
# 일관성지수 공식 (고윳값-n)/(n-1)/무작위지수
# 무작위 지수(RI)는 계산된 표가 존재함
CR = ((np.real(val[0]) - 3) / 2) / 0.58
# CR
print('가중치 :', w.round(3), ' CR :', round(CR, 3))
```

```python
# 첫 열과 첫행에 필요한 값이 있었지만 아닐수도 있다. 이경우 아래 함수를 써서 유연하게 대응이 가능하다.
# argmax() : 가장 큰 값의 인덱스
# max() 가장 큰 값 
w = np.real(vec[:, val.real.argmax()]) 
w /= w.sum()

CR = ((val.real.max() - 3) / 2) / 0.58

print('가중치 :', w.round(3), ' CR :', round(CR, 3))
```
