```python
import numpy as np
import sklearn.model_selection as ms
import sklearn.utils as util
from sklearn.linear_model import Ridge
from sklearn.linear_model import Lasso

arr = np.load('/content/drive/MyDrive/운영/MBC/데이터분석_1기/수업/04_ML/data/boston.npz')
x = arr['x']
y = arr['y']
```

```python
alphas = [0.001,0.01,0.1,1,10]
iters = [1000,5000,10000,50000,100000]

util.shuffle(x,y,random_state=777) #

#Rige 모델 최적의 파라메터 조합과 점수
params = {'max_iter':iters,'alpha':alphas} # 모델에서 사용하는 파라메터이름 : 그 값들
grid = ms.GridSearchCV(Ridge(), param_grid=params, cv=5).fit(x,y)
print(f'Ridge Best Param : {grid.best_params_}')
print(f'Ridge Best Score : {grid.best_score_}')

#Lasso 모델 최적의 파라메터 조합과 점수
iters = [10000,50000,100000,500000,1000000]
params = {'max_iter':iters,'alpha':alphas} # 모델에서 사용하는 파라메터이름 : 그 값들
grid = ms.GridSearchCV(Lasso(), param_grid=params, cv=5).fit(x,y)
print(f'Lasso Best Param : {grid.best_params_}')
print(f'Lasso Best Score : {grid.best_score_}')
```

CV(Cross Validation) - 교차검증
<p>cv=5 가 의미하는 것</p>
1. 주워진 데이터를 동등한 크기로 5분할 한다.<br/>
2. 모델 훈련과 평가를 5번 수행한다.<br/>
3. 각 평가마다 4개의 SET으로 훈련, 나머지 1개 SET으로 테스트 진행<br/>
4. 해당 조합의 성능을 5번 진행하는동안의 평균을 계산하여 최고치를 출력<br/>
