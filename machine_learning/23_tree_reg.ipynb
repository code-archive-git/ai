```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor

#colab -> Colab CPU -> Open terminal -> mkdir data -> cd data
# github 의 raw data 주소 확인
# wget https://github.com/code-archive-git/ai/raw/refs/heads/main/machine_learning/data/make_wave.npz
# 1. 데이터 불러오기
arr=np.load('/content/data/make_wave.npz')
x = arr['x']
y = arr['y']

# 2. 시각화로 데이터 파악
plt.scatter(x,y)
plt.show()
```

```python
# 3. 데이터 분리 -> 샘플 수가 너무 적어 분리작업 생략
# 4. 가지고 있는 데이터로 학습
tree = DecisionTreeRegressor(max_depth=3).fit(x,y)

# 5. 테스트 - 새로만든 데이터로 predict(예측) 시켜 본다.
# -3 ~ 3 까지 0.01 단위를 x 값으로 해서 예측하도록 시킨다.
x_test = np.arange(-3,5,0.01).reshape(-1,1) # 질문데이터는 반드기 아파트(2차원 배열)형태로
y_result = tree.predict(x_test)

# 그 결과값 y 를 이용해 어떤 패턴으로 예측하는지 라인으로 이어본다.
plt.plot(x_test,y_result, c='red', label='predict')
plt.scatter(x,y,c='blue', label='train_data')
plt.legend()
plt.show()
# 회귀 트리는 학습한 데이터 영역 밖은 제대로 예측해주지 못한다.
```
