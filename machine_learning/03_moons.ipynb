```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
# make_moons 는 초승달 모양의 클러스터 데이터를 생성
# 여기에서 생성된 데이터는 직선을 활용해서 분류할 수 없다.
plt.subplot(1,2,1)
x,y=make_moons(n_samples=400, noise=0)
plt.scatter(x[:,0],x[:,1],c=y)
plt.title('noise=0')

plt.subplot(1,2,2)
x,y=make_moons(n_samples=400, noise=1)
plt.scatter(x[:,0],x[:,1],c=y)
plt.title('noise=1')

plt.show()
```

n_samples : 몇개의 데이터 추출?
<br/>
noise : 0~1, 0이면 정확한 반원을 이룸
