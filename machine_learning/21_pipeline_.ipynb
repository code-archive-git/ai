```python
from sklearn.datasets import fetch_20newsgroups
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
```

```python
# 1. 뉴스그룹
#news = fetch_20newsgroups(subset='all')
news = fetch_20newsgroups(
    data_home='/content/drive/MyDrive/Lecture/AI/04_SCIKIT_LEARN/data',
    subset='all',
    download_if_missing=False)
print(news.keys())
print(len(news.target_names))

# 2. 데이터 할당
x = news.data
y = news.target
print(x[0])
print(f'{y[0]} - {news.target_names[y[0]]}')
```

```python
# 데이터 분리
train_x, test_x, train_y, test_y = train_test_split(x, y, test_size=0.3, random_state=777)

# pipe line 활용
# 사용할 스케일 함수 지정 -> fit_transform 으로 스케일 실행 -> 사용할 모델 지정 -> train -> 점수 산정
from sklearn.pipeline import Pipeline
import sklearn.feature_extraction.text as text

# 사용할 스케일 함수 지정 -> fit_transform 으로 스케일 실행 -> 사용할 모델 지정
model1 = Pipeline([
  ('vect',text.CountVectorizer()),
  ('model', MultinomialNB())
])

model2 = Pipeline([
  ('vect',text.TfidfVectorizer()),
  ('model', MultinomialNB())
])

model3 = Pipeline([
  ('vect',text.TfidfVectorizer(stop_words='english')),
  ('model', MultinomialNB())
])

# for i in list : <- 뒤에있는 리스트 값을 하나씩 빼서 앞으로 전달
# for a,b in zip(list1,list2) <- 뒤에있는 list1, list2 에서 값을 각각 빼내, 앞에있는 a와 b에게 전달
# for i, model in enumerate(list) <- 뒤에있는 리스트 값을 하나씩 빼서 인덱스와 함께 앞으로 전달

for i,model in enumerate([model1, model2, model3]):
  model.fit(train_x,train_y)
  print(f'model-{i} : {model.score(test_x, test_y)}')

```

```python

result = model3.predict(["hello world"])
result
print(f'{result} - {news.target_names[result]}')
```
