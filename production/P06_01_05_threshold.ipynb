```python
import numpy as np

rng = np.random.default_rng(0)

# 클수록 이상한 점수(Anomaly Score)를 직접 생성
s_norm = rng.normal(loc=0.0, scale=1.0, size=300)  # 정상 데이터 300개의 이상 점수 (낮은 분포)
s_out = rng.normal(loc=2.5, scale=1.0, size=50)  # 이상 데이터 50개의 이상 점수 (높은 분포)

for p in [90, 95, 99]:
  thr = np.percentile(s_norm, p)  # 정상 점수 하위 p% 지점을 임계값(thr)으로 자동 계산
  print(
      f"{p}% 임계 (thr={thr:.2f}) : 재현율 {(s_out > thr).mean():.2f}, 오탐률"
      f" {(s_norm > thr).mean():.2f}"
  )
```
