```python
from sklearn.datasets import fetch_20newsgroups

# 1. 데이터 호출
news = fetch_20newsgroups(subset='all') # train, test, all
print(news.keys())
print(news.target_names) # 뉴스 분류 명
```

```python
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
import sklearn.feature_extraction.text as text
# 2. 데이터 분리
x_train, x_test, y_train, y_test = train_test_split(news.data, news.target, test_size=0.3, random_state=777)

#  벡터화(컴퓨터가 이해할수 없는 데이터 를 이해할(계산할) 수 있도록 변경하는 작업)
# 3.벡터함수 호출 - 단어 출현 빈도를 수치화

# CountVectorizer : 문자열 데이터 빈도를 수치화 시킨다.
# 단점 : 의미가 없다 해도 자주 등장하면 가중치가 높아진다.
counter = text.CountVectorizer()

# 데이터를 훈련(fit) 하기 위한 벡터화(fit_transform)
vx = counter.fit_transform(x_train)

# print(vx)
# Coords	    Values
# (0, 62688)	  1

# 알고리즘 학습
multi = MultinomialNB().fit(vx,y_train)

# 결과 확인 - 결과 확인시 벡터화(transform)
vx_test = counter.transform(x_test)
print(f'test : {multi.score(vx_test,y_test)}')
```

```python
# Counter 의 단점을 보완하기 위해 Tfidfvectorizer() 등장
# 특정단어가 몇개의 문장에서 나왔는지 수치화 하여 계산한다.
tfid = text.TfidfVectorizer()
vx_train = tfid.fit_transform(x_train)
vx_test = tfid.transform(x_test)

multi = MultinomialNB().fit(vx_train,y_train)
print(f'train : {multi.score(vx_train,y_train)}')
print(f'test : {multi.score(vx_test,y_test)}')

```
