```python
# 근접률 = |값 − SV|/(공차/2)
# 근접률 = 0: 정중앙(SV)에 완벽하게 일치함 (가장 이상적)
# 근접률 = 1: 규격 한계선(USL 또는 LSL)에 아슬아슬하게 걸쳐 있음 (불량 직전)
# 근접률 > 1: 규격 범위를 벗어남 (불량 발생!)
```

```python
SV, LSL, USL = 10.00, 9.95, 10.05

for val in [10.00, 10.03, 10.045, 10.06]:
    rate = abs(val - SV) / ((USL - LSL) / 2)
    print(f'{val} 근접률 = {rate:.1f}')
```
