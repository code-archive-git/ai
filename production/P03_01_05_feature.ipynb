```python
import pandas as pd

s = pd.Series([180, 181, 183, 182, 185, 186, 184, 187], name='temp')
fe = pd.DataFrame({'temp': s, 'lag1': s.shift(1), 'lag3': s.shift(3),
                   'ma3': s.rolling(3).mean(), 'diff': s.diff()})
```

```python
print(fe)
# lag1 : 1스텝 이전값, 처음엔 이전값이 없으므로 NaN 발생
# lag3 : 3스텝 이전값 
# ma3 : 3개 이동평균 (인덱스 2의 값은 (183+181+180)/3 = 181.33).
# diff : 현재 값에서 바로 직전 값을 뺀 값(공정 데이터가 상승세인지 하락세인지 트렌드 파악)
```
