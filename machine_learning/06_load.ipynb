```python
from joblib import load # dump 로 저장한 내용을 load 로 읽어들인다.
path = '/content/drive/MyDrive/운영/MBC/데이터분석_1기/수업/04_ML/Data/knn_iris.joblib'

names = ['Setosa','Virsicolor','Virginica']

# 저장된 내용 불러오기
class_knn = load(path)

sample = [[5,5,3,2],[2,4,3,5]]

results = class_knn.predict(sample)

for r in results:
  print(names[r])
```
