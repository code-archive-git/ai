```python
import numpy as np

rng = np.random.default_rng(0)

# 클수록 이상한 점수(Anomaly Score)를 직접 생성
s_norm = rng.normal(loc=0.0, scale=1.0, size=300)  # 정상 데이터 300개의 이상 점수 (낮은 분포)
s_out = rng.normal(loc=2.5, scale=1.0, size=50)  # 이상 데이터 50개의 이상 점수 (높은 분포)

for p in [90, 95, 99]:
  # 정상 데이터(s_norm)을 오름차순으로 하여, 하위 p% 위치에 해당하는 경계를 구해 임계값(threshold)으로 지정
  thr = np.percentile(s_norm, p)

  # (s_out|s_norm) 이 thr 기준을 얼마나 넘어갔나? [True, False, ..] 의 비율을 구한다.
  # 재현율:s_out > thr -> s_out 이 thr 기준으로 벗어난 비율(이상치 발견 확률)
  # 오탐률:s_norm > thr -> s_norm 이 thr 기준으로 벗어난 비율(정상을 이상치라고 오해한 비율)
  print(
      f"{p}% 임계 (thr={thr:.2f}) : 재현율 {(s_out > thr).mean():.2f}, 오탐률"
      f" {(s_norm > thr).mean():.2f}"
  )
```
