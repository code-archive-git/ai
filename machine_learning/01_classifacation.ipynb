scikit-learn 에서는 분류모형 테스트를 위한 데이터 생성 함수를 제공한다.

* n_features=1 : 입력값에 포함되어야 할 전체 특성(분류기준)의 수, 기본 20개 카테고리
* n_classes=2 : 분류되어야항 class 의 수, 기본 2개(예: 스팸/정상, 합격/불합격)
* n_informative=1 : 전체 입력값 중에  실제로 구분하는데 유용한 정보를 가진 특성의 수(기본 2개)
* n_clusters_per_class=1 : 하나의 클래스내에 데이터가 뭉쳐있는 군집의 수(기본 2개)
* n_redundant=0 :유용한 정보를 가진 특성들을 선형조합하여 만든 중복된 특성의 수(기본 2개)



```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification

x,y=make_classification(n_features=1, n_classes=2, n_informative=1, n_clusters_per_class=1, n_redundant=0)
#print(x) # 독립변수
#print(y) # 종속변수
plt.scatter(x,y, c=y)
plt.yticks([0,0.5,1])
plt.show()
```

n_samples= 보여줄 샘플 갯수 지정

```python
x,y=make_classification(n_features=2, n_classes=2, n_informative=1, n_clusters_per_class=1, n_redundant=0, n_samples=400)
plt.scatter(x[:,0],x[:,1],c=y)
plt.show()
```

```python
x,y=make_classification(n_features=2, n_classes=2, n_informative=2,
                        n_clusters_per_class=1, n_redundant=0, n_samples=500)
plt.scatter(x[:,0],x[:,1],c=y)
plt.show()
```

```python
# 각 클래스의 비율을 조정하고 싶을 때
x,y = make_classification(n_features=2, n_classes=2, n_informative=2,
    n_clusters_per_class=1, n_redundant=0, n_samples=1000, weights=[0.9,0.1])
plt.scatter(x[:,0],x[:,1],c=y)
```

```python
# n_clusters_per_class 는 클래스의 군집 중심의 갯수 - 많아질수록 데이터 분리가 어려워진다.
x,y = make_classification(n_features=2, n_classes=3, n_clusters_per_class=2,
                          n_informative=2, n_redundant=0, n_samples=500)

# 수치 관계
# n_classes * n_clusters_per_class <= 2 ** n_informative
# n_features >= n_informative + n_redundant + n_repeated

plt.scatter(x[:,0],x[:,1],c=y)
plt.show()
```

```python
# class 3개로 분류되는 데이터 만들어 보기
x,y = make_classification(n_features=3, n_classes=3, n_informative=3, n_samples=500,
    n_clusters_per_class=1, n_redundant=0)

print(f'cmap  종류 : {plt.colormaps()}')

plt.scatter(x[:,0],x[:,1],c=y, s=(x[:,2]*30), cmap='plasma')
plt.show()
```
