```python
import pandas as pd
# 1. 샘플 협력사 데이터 (예: 리스크 관점 원시값)
# '품질': 높을수록 불량·리스크 큼, '납기': 높을수록 지연·리스크 큼, '단가': 높을수록 비용 부담·리스크 큼
df = pd.DataFrame({
    "협력사": ["S01", "S02", "S03", "S04", "S05", "S06"],
    "품질": [0.5, 2.0, 1.0, 3.5, 1.2, 4.0],
    "납기": [0.20, 0.05, 0.10, 0.30, 0.15, 0.25],
    "단가": [100, 90, 110, 120, 95, 130],
})

# 2. AHP로 산출한 가중치 (합계 = 1.0)
w = {'품질': 0.648, '납기': 0.23, '단가':0.122}
```

```python
for k in w:
    # minMax 정규화로 0~1 사이 점수로 정규화(이번엔 작을수록...)
    df[k + '_n'] = (df[k] - df[k].min()) / (df[k].max() - df[k].min())
df # 정규화 한 내용
```

```python
# 정규화 점수를 합산한다.
df['risk'] = 0.0
for k, v in w.items():
    # 각 컬럼에 가중치를 곱한다음 합하여 risk 에 담는다.
    df['risk'] += df[k+'_n'] * v
df['risk'] = round(df['risk'] * 100,1) # 100점 만점 표현을 위해 100을 곱한다.
df.sort_values('risk')
```

```python
# risk 만 잘라서 cut 과 qcut 으로 등급화 해 본다.
df['cut'] = pd.cut(df['risk'],bins=[0,30,50,70,100], labels=["A","B","C","D"])
df['qcut'] = pd.qcut(df['risk'], q=4, labels=['A', 'B', 'C', 'D'])
df.sort_values('risk')
```
