```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
# mean_absolute_error: MAE(평균 절대 오차) 계산용
# mean_squared_error: MSE(평균 제곱 오차) 계산용 (여기에 루트를 씌우면 RMSE)
# r2_score: R²(결정계수) 계산용

y_true = [10.00, 10.02, 9.98, 10.01, 9.99]
y_pred = [10.01, 10.01, 9.99, 10.03, 9.98]

# 평균적으로 얼마나 틀리는가
print('MAE  :', mean_absolute_error(y_true, y_pred))
# 가끔 크게 틀리는 오차를 매섭게 경계
print('RMSE :', np.sqrt(mean_squared_error(y_true, y_pred)))
# 내 모델이 데이터를 전체적으로 얼마나 찰떡같이 잘 설명하는지
print('R2   :', r2_score(y_true, y_pred))
```
