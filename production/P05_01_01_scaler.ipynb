```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler

x = np.array([[1.0], [1.2], [0.9], [1.1], [8.0]])     # 불량률(%), 마지막 8.0은 이상치
result = pd.DataFrame({'origin': x.ravel()})

# 각기 다른 스케일러 적용
scales = [('MinMax', MinMaxScaler()), ('Standard', StandardScaler()), ('Robust', RobustScaler())]
for name, sc in scales:
    result[name] = sc.fit_transform(x).ravel().round(2)
# 결과
result
```

```python
import matplotlib.pyplot as plt
labels = ['S01', 'S02', 'S03', 'S04', 'S05']
for name, data in result.items():
    if name == 'origin':
        plt.bar(labels,data, color='gray', alpha=0.5)
    else:
        plt.plot(labels, data, label=name, linewidth=2, markersize=8)
        plt.spring()
        plt.legend(fontsize=11)
        plt.grid(True, linestyle='--', alpha=0.6)
        plt.axhline(0, color='gray', linewidth=0.8, linestyle='--')
plt.show()
```
