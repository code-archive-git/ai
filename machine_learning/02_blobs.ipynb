```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
```

make_blobs : 등방성 가우시안 정규분포 를 이용해 데이터 생성
<br/>
등방성 : 같은 방향으로 나가는 성질

```python
x,y = make_blobs(n_samples=300, n_features=2, centers=2, cluster_std=1)
plt.subplot(121)
plt.scatter(x[:,0],x[:,1],c=y)
plt.subplot(122)
x,y = make_blobs(n_samples=300, n_features=2, centers=3, cluster_std=3)
plt.scatter(x[:,0],x[:,1],c=y)
plt.show()
```

n_samples= 데이터 수
<br/>
n_features=입력한 독립변수
<br/>
centers=결과로 나타날 군집 수
<br/>
cluster_std=클러스터의 표준 편차. 값이 클수록 데이터가 넓게 퍼짐
