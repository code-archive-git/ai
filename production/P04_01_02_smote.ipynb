```python
!pip install imbalanced-learn
from collections import Counter
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import SMOTE
```

```python
# [0]양품 95% / [1]불량 5% 데이터 생성
x, y = make_classification(n_samples=2000, n_features=5, n_informative=3,
                           weights=[0.95, 0.05], random_state=0)
print(Counter(y))   # {0: 1892, 1: 108} 정도
```

```python

# stratify=y : 불균형 비율(95:5)이 학습용과 테스트용 세트에 똑같이 유지되도록 쪼개주는 옵션
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, stratify=y, random_state=0)

smote = SMOTE(random_state=0) # SMOTE 객체 생성

# 학습(train) 데이터에만 사용 한다.
# test 까지 SMOTE를 쓰면 합성된 가짜 데이터가 테스트셋에 섞여 들어가 평가 점수가 왜곡 된다.
x_res, y_res = smote.fit_resample(x_train, y_train)
print(Counter(y_res))   # 두 클래스 수가 95%, 5% 에서 50%, 50% 로 같아짐
```
