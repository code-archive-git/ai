```python
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression

iris = load_iris()
model = LogisticRegression(max_iter=1000).fit(iris.data, iris.target)

sample = iris.data[[0, 60, 120]] # iris data 중 0번 60번 120번 인덱스에 있는 데이터만 가져옴
print(model.predict_proba(sample).round(3))# predict_proba : 입력 샘플이 각 클래스에 속할 확률
print(iris.target_names[model.predict(sample)])   # 가장 높은 확률의 클래스
```
